```python
def get_idos_barycentric_2(pot_avg, pot_diff, pot_min, pot_max, en_min, en_max):


        restrict_domain_2(pot_diff, pot_min, pot_max)
        restrict_domain_2(pot_avg, en_min, en_max)


    
        return idos_interpolated(pot_diff, pot_avg)

def restrict_domain_2(array, min_val, max_val):
        ### numpy fancy idexing

        #https://stackoverflow.com/questions/19666626/replace-all-elements-of-numpy-array-that-are-greater-than-some-value

        array[array < min_val] = min_val
        array[array > max_val] = max_val

        return
```
